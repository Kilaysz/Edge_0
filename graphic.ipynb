## Grafik Hasil Skenario 1 - Iterasi

```python
!pip install pandas matplotlib
import pandas as pd
import matplotlib.pyplot as plt

# Baca file CSV hasil skenario 1
df = pd.read_csv('results/csv/skenario1.csv')

# Cek data awal
print(df.head())

# Buat plot
plt.figure(figsize=(8, 5))

# Plot nilai rata-rata terhadap jumlah iterasi
plt.plot(df['max_iter'], df['Average'], marker='o', linestyle='-', color='steelblue')

# Tambahkan nilai di atas tiap titik
min_y = df['Average'].min()  # cari nilai minimum

for x, y in zip(df['max_iter'], df['Average']):
    if y == min_y:
        plt.text(x, y, f"{y:.5f}", ha='center', va='bottom', fontsize=8, fontweight='bold')
    else:
        plt.text(x, y, f"{y:.5f}", ha='center', va='bottom', fontsize=8)

# Label dan judul
plt.xlabel('Jumlah Iterasi')
plt.ylabel('Rerata Fitness Gabungan')
plt.title('Pengaruh Jumlah Iterasi terhadap Fitness')
plt.grid(True)

# Tampilkan plot
plt.tight_layout()

#Simpan Grafik
plt.savefig('results/graphic/skenario1.png', dpi=140) 

plt.show()
```

## Grafik Hasil Skenario 5 - Server

```python
import pandas as pd
import matplotlib.pyplot as plt

# Baca data
df = pd.read_csv('results/csv/skenario5.csv')

# Filter hingga K <= 500
df = df[df['K'] <= 500].copy()

# Hitung presentase penurunan dibandingkan sebelumnya
df['Drop (%)'] = df['Average'].shift(1) - df['Average']
df['Drop (%)'] = (df['Drop (%)'] / df['Average'].shift(1)) * 100

# Plot
fig, ax1 = plt.subplots(figsize=(10, 6))

# Plot garis untuk rata-rata fitness
color1 = 'tab:blue'
ax1.set_xlabel('Jumlah Server (K)')
ax1.set_ylabel('Fitness Gabungan', color=color1)
ax1.plot(df['K'], df['Average'], marker='o', color=color1, label='Rata-rata Fitness')
ax1.tick_params(axis='y', labelcolor=color1)
ax1.grid(True)

# Sumbu y kedua untuk presentase penurunan
ax2 = ax1.twinx()
color2 = 'tab:red'
ax2.set_ylabel('Penurunan (%)', color=color2)
ax2.bar(df['K'], df['Drop (%)'], color=color2, alpha=0.3, label='Penurunan Fitness')
ax2.tick_params(axis='y', labelcolor=color2)

# Judul dan layout
plt.title('Pengaruh Jumlah Server Terhadap Fitness')
fig.tight_layout()

# Simpan ke file
plt.savefig('results/graphic/skenario5.png', dpi=140)
plt.show()
```

## Grafik Hasil Skenario 6 - QPSO Non PU vs QPSO PU

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load hasil skenario 6
df = pd.read_csv('results/csv/skenario6.csv')

# Konversi kolom numerik
df['Average'] = df['Average'].astype(float)
df['avg_distance'] = df['avg_distance'].astype(float)
df['workload_var'] = df['workload_var'].astype(float)
df['filter_percent'] = df['filter'].str.replace('%', '').astype(int)


df_pu = df[df['filter_percent'] == 50].copy()
df_nonpu = df[df['filter_percent'] == 0].copy()

plt.figure(figsize=(16, 10))

# Plot 1: Jarak Rata-rata
plt.subplot(2, 2, 1)
plt.plot(df_nonpu['num_edge_servers'], df_nonpu['avg_distance'], marker='o', color='gray', label='Tanpa PU')
plt.plot(df_pu['num_edge_servers'], df_pu['avg_distance'], marker='o', color='darkorange', label='Dengan PU (50%)')
plt.title('Perbandingan Jarak Rata-rata')
plt.xlabel('Jumlah Edge Server')
plt.ylabel('Jarak Rata-rata (km)')
plt.xticks(df_nonpu['num_edge_servers'])
plt.grid(True)
plt.legend()

# Plot 2: Varians Beban Kerja
plt.subplot(2, 2, 2)
plt.plot(df_nonpu['num_edge_servers'], df_nonpu['workload_var'], marker='o', color='gray', label='Tanpa PU')
plt.plot(df_pu['num_edge_servers'], df_pu['workload_var'], marker='o', color='darkorange', label='Dengan PU (50%)')
plt.title('Perbandingan Varians Beban Kerja')
plt.xlabel('Jumlah Edge Server')
plt.ylabel('Standar Varians Beban (menit)')
plt.xticks(df_nonpu['num_edge_servers'])
plt.grid(True)
plt.legend()

# Plot 3: Fitness Gabungan
plt.subplot(2, 2, 3)
plt.plot(df_nonpu['num_edge_servers'], df_nonpu['Average'], marker='o', color='gray', label='Tanpa PU')
plt.plot(df_pu['num_edge_servers'], df_pu['Average'], marker='o', color='darkorange', label='Dengan PU (50%)')
plt.title('Perbandingan Fitness QPSO')
plt.xlabel('Jumlah Edge Server')
plt.ylabel('Fitness Gabungan')
plt.xticks(df_nonpu['num_edge_servers'])
plt.grid(True)
plt.legend()

# Tata letak dan simpan
plt.tight_layout()
plt.savefig('results/graphic/skenario6.png', dpi=140)
plt.show()
```

## Grafik Hasil Skenario 7 - Pembanding Non PU

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load hasil skenario 7
df = pd.read_csv('results/csv/skenario7.csv')

# Pastikan urutan algoritma konsisten
algorithms = ['QPSO', 'K-means', 'Top-K', 'Random', 'WeightedKMeans']
colors = {
    'QPSO': 'darkorange',
    'K-means': 'royalblue',
    'Top-K': 'green',
    'Random': 'rebeccapurple',
    'WeightedKMeans': 'orangered'
}

# Urutan jumlah edge server
k_values = sorted(df['num_edge_servers'].unique())

# Siapkan plot
plt.figure(figsize=(16, 12))

# Plot 1: Rata-rata Jarak (avg_distance)
plt.subplot(2, 2, 1)
for alg in algorithms:
    subset = df[df['placer_name'] == alg].sort_values('num_edge_servers')
    if not subset.empty:
        plt.plot(subset['num_edge_servers'], subset['avg_distance'], marker='o', label=alg, color=colors.get(alg, 'black'))
plt.title('Metrik Jarak')
plt.xlabel('Jumlah Edge Server')
plt.ylabel('Jarak Rata-rata (km)')
plt.grid(True)
plt.legend()

# Plot 3: Varians Beban Kerja (workload_var)
plt.subplot(2, 2, 2)
for alg in algorithms:
    subset = df[df['placer_name'] == alg].sort_values('num_edge_servers')
    if 'workload_var' in subset.columns and not subset.empty:
        plt.plot(subset['num_edge_servers'], subset['workload_var'], marker='o', label=alg, color=colors.get(alg, 'black'))
plt.title('Metrik Beban Kerja')
plt.xlabel('Jumlah Edge Server')
plt.ylabel('Varians Beban Kerja (menit)')
plt.grid(True)
plt.legend()

# Plot 4: Fitness Gabungan
plt.subplot(2, 2, 3)
for alg in algorithms:
    subset = df[df['placer_name'] == alg].sort_values('num_edge_servers')
    if not subset.empty:
        plt.plot(subset['num_edge_servers'], subset['fitness'], marker='o', label=alg, color=colors.get(alg, 'black'))
plt.title('Fitness Gabungan')
plt.xlabel('Jumlah Edge Server')
plt.ylabel('Fitness')
plt.grid(True)
plt.legend()

# Tata letak akhir
plt.tight_layout()
plt.savefig('results/graphic/skenario7.png', dpi=100)
plt.show()
```
