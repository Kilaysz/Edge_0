## 1. Import Library

```python
import pandas as pd
import numpy as np
import os
```

```python
# Direktori file dataset
files_dir = '../dataset/raw'

# Ambil semua nama file .xlsx
file_names = [fname for fname in os.listdir(files_dir) if fname.endswith('.xlsx')]

# Buat path lengkap tiap file
file_paths = [os.path.join(files_dir, fname) for fname in file_names]

# Baca semua file Excel dan simpan ke list
raw_data_list = []
for fpath in file_paths:
    raw_data = pd.read_excel(fpath)
    raw_data_list.append(raw_data)

# Gabungkan semua DataFrame jadi satu
raw_data = pd.concat(raw_data_list)

# Simpan hasil gabungan ke file CSV
raw_data.to_csv('../dataset/raw_data.csv', index=False)
```

## 4. Preprocess Data

```python
# Import data mentah dari file CSV
raw_data = pd.read_csv('../dataset/raw_data.csv')

# Konversi kolom waktu ke format datetime
raw_data['start time'] = pd.to_datetime(raw_data['start time'])
raw_data['end time'] = pd.to_datetime(raw_data['end time'])

# Urutkan data berdasarkan waktu mulai
clean_data = raw_data.sort_values(by='start time')

# Hapus baris yang mengandung nilai kosong
clean_data = clean_data.dropna()

# Buat kolom 'address' dari gabungan latitude dan longitude
clean_data['address'] = clean_data.apply(lambda x: f"{x['latitude']}-{x['longitude']}", axis=1)

# Simpan data yang telah dibersihkan
clean_data.to_csv('../dataset/data_all.csv', index=False)

# Ekstrak data base station unik berdasarkan lokasi
bs_data = clean_data[['latitude', 'longitude', 'address']].drop_duplicates()

# Reset index dan beri ID numerik
bs_data = bs_data.reset_index(drop=True)
bs_data['id'] = bs_data.index

# Susun ulang kolom agar 'id' di depan
bs_data = bs_data[['id', 'latitude', 'longitude', 'address']]

# Simpan data base station ke file CSV
bs_data.to_csv('../dataset/bs_all.csv', index=False)
```
